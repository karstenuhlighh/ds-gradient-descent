## Der Algorithmus

Ein einfacher Gradientenabstiegs-Algorithmus sieht wie folgt aus:

1. Erhalte eine zu minimierende Funktion F(x)

2. Initialisiere einen Wert x, von dem aus der Abstieg oder die Optimierung gestartet wird

3. Spezifiziere eine Lernrate, die bestimmt, wie groß ein Schritt beim Abstieg ist oder wie schnell du zum Minimalwert konvergierst

4. Erhalte die Ableitung dieses Wertes x (der Abstieg)

5. Fahre fort, um die Ableitung dieses Wertes multipliziert mit der Lernrate abzusteigen

6. Passe den Wert von x an

7. Überprüfe deine Abbruchbedingung, um zu sehen, ob du stoppen sollst

8. Wenn die Bedingung erfüllt ist, stoppe. Wenn nicht, fahre mit Schritt 4 mit dem neuen x-Wert fort und wiederhole den Algorithmus weiter
